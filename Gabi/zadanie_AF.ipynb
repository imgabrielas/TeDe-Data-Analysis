## Zadanie 3: Brakujące wartości
Policz braki w każdej kolumnie. Następnie uzupełnij brak w kolumnie `punkty` średnią z tej kolumny.

**Podpowiedź:**
```python
df.isna().sum()                                  # liczba braków w kolumnach
df['kolumna'] = df['kolumna'].fillna(wartosc)    # wstaw wartość w miejsce braków
df['kolumna'].mean()                             # średnia z kolumny
```

````python
# Twój kod tutaj
df.isna().sum()
````

````python
df['punkty'] = df['punkty'].fillna(df['punkty'].mean())
df.head(50)
````

## Zadanie 5: Dodaj nową kolumnę
Dodaj kolumnę `wynik`, w której będzie `'zdał'`, jeśli student ma co najmniej 60 punktów, albo `'nie zdał'` w przeciwnym razie.

**Podpowiedź (NumPy):**
```python
df['nowa_kolumna'] = np.where(warunek, 'jeśli prawda', 'jeśli fałsz')
```

````python
# Twój kod tutaj
df['wynik'] = np.where(df['punkty'] >= 60, 'zdal', 'nie zdal')
df.head(50)
````

## Zadanie 6: Posortuj dane

## Zadanie 8: Wykres w Matplotlib
